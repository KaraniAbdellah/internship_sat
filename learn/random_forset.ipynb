```python
# Introduction
'''
    One of The big Problems in Descion Tree is Overfitting OR underfitting
    Random Forset Solution: this also solve the problem of ....
'''
```

```python
# Import Packages
import pandas as pd
import seaborn as sns
```

```python
# Get All Dataset Predifined
sns.get_dataset_names()
```

```python
healthexp = sns.load_dataset("healthexp")
healthexp.head(3)
```

```python
# Convert categorical variable into dummy/indicator variables. 
healthexp = pd.get_dummies(healthexp)
healthexp.head(2)
```

```python
# Get X and y
X = healthexp.drop(["Life_Expectancy"], axis=1)
y = healthexp["Life_Expectancy"]
X.head(2)
```

```python
# Split Test Set and Train Set
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2) 
y_test.head(3)
```

```python
# Build Random Forset Model
from sklearn.ensemble import RandomForestRegressor
# Build Model With Set Important Parameter to Avoid Overfitting
rfm = RandomForestRegressor(random_state=0, n_estimators=100, oob_score=True)
```

```python
# Fit X and y Train
rfm.fit(X_train, y_train)
```

```python
# Make Prediction
y_prediction = rfm.predict(X_test)
print("Test Set: ", y_test.to_list())
print("Prediction Set: ", y_prediction)
```

```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
print(mean_absolute_error(y_prediction, y_test))
print(root_mean_squared_error(y_prediction, y_test))
print(r2_score(y_prediction, y_test))
```

#### OOB Error (out-of-bug Error)

```python
print(f"OOB R-squared: {rfm.oob_score_}") # great prediction it close to 1
```
