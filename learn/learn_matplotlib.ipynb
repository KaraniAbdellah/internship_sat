## SubPlot

```python
# SubPlot: group of plots
fig, axs = plt.subplots(2, 2, figsize=(12, 10)) # number of plots horizentally and vertically
axs[0, 0].hist([1, 2, 3], bins = 10, color="lightblue")
axs[0, 1].scatter([1, 2], [3, 4], color="purple")
axs[1, 0].bar([1, 2], [3, 4], color="green")
axs[1, 1].plot([1, 2], [3, 4], color="red")
fig.show()
```

```python
df_1 = cp.deepcopy(df)
fig, axs = plt.subplots(2, 2, figsize=(15, 10)) # figsize=(width, height)

# Price vs. Frequency
axs[0, 0].hist(df_1["price"], bins=20, color="green")
axs[0, 0].set_title("Price vs. Frequency")
axs[0, 0].set_xlabel("Price")
axs[0, 0].set_ylabel("Frequency")

# Group by type of room and show average price
df_p_r = pd.concat([df_1["price"], df_1["room_type"]], axis=1)
grouped_df_p_r = df_p_r.groupby("room_type")["price"].sum() # sum: get sum of price columns
axs[0,1].bar(grouped_df_p_r.index, grouped_df_p_r.values)
axs[0,1].set_title("Price Sum By Room Type")
```

##### Stacked bar Chart

```python
import matplotlib.pyplot as plt
category = ["A", "B", "C"]
val1 = [20, 32, 34]
val2 = [15, 20, 12]
val3 = [10, 15, 8]

plt.bar(category, val1, label="plot1")
plt.bar(category, val2, label="plot2")
plt.bar(category, val3, label="plot3")
plt.title("Stacked Bar Chart")
plt.ylabel("values")
plt.xlabel("catagory")
plt.legend()
plt.show()

```

```python
genres = ['Comedy', 'Action', 'Darama', 'Scifi', 'romantic']
boys = [30, 20, 40, 50, 15]
girls = [20, 40, 20, 15, 35]

# Stacked Bar Graphs
# plt.bar(genres, boys, label="boys", bottom=girls)
# plt.bar(genres, girls, label="girls")
# plt.xlabel("Genres")
# plt.ylabel("Girsl/Boys")
# plt.title("Favorite Type of Movies")
# plt.legend()
# plt.show()

# Multiple Bar Graphs
values_genres = np.arange(len(genres))
print(values_genres)

width = 0.4
plt.subplots(figsize=(16, 12))
plt.bar(values_genres, boys, width=width, label="BOYS")
plt.bar(values_genres + width, girls, width=width, label="GiRLS")

plt.legend()
plt.xticks(values_genres + (width / 2), genres)
plt.show()
```

#### Pivot Table

```python
import pandas as pd
# Create DataFrame
df = pd.DataFrame({
    "Region": ["North", "South", "North", "South"],
    "Product": ["Apples", "Oranges", "Oranges", "Apples"],
    "Sales": [100, 150, 200, 120]
})

df
df_pivot = df.pivot_table(index="Region", columns="Product", values="Sales")
df_pivot

df_pivot.plot(kind="bar")
```
